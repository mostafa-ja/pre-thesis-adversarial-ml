# Converting the Drebin features to sparse matrices

A small utility notebook with an outsized role: the detectors do not read the raw Drebin pickles — they load **sparse matrices** (`sparse_matrix_0.npz` and its siblings). This is the conversion that produced them: load the pickled feature data, convert to CSR sparse format, and save as `.npz`.

*Consolidated from `Untitled1.ipynb`, a scratch notebook found inside the `drebin/` folder of `mal_adv3`. Duplicated shape-check cells removed; outputs cleared.*

## Step 1 — Load the pickled features

```python
import sklearn.externals.joblib
import joblib
import pickle
with open(r"Desktop/X.pkl", "rb") as f:
    data = joblib.load(f)
```

```python
len(data)
```

```python
data[0].shape
```

## Step 2 — Convert to sparse (CSR) format

```python
from scipy import sparse
# Convert it to a sparse matrix format, for example, CSR (Compressed Sparse Row)
sparse_matrix_0 = sparse.csr_matrix(data[0])
sparse_matrix_1 = sparse.csr_matrix(data[1])
sparse_matrix_2 = sparse.csr_matrix(data[2])
```

```python
from scipy import sparse

# Assuming your sparse matrix is named sparse_matrix
sparse.save_npz("sparse_matrix2_0.npz", sparse_matrix_0)
sparse.save_npz("sparse_matrix2_1.npz", sparse_matrix_1)
sparse.save_npz("sparse_matrix2_2.npz", sparse_matrix_2)
```
